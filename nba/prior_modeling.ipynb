```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
```

```python
nba_training_df = pd.read_csv('data/training_games4.csv')
nba_training_df.head()
```

```python
nba_training_df.drop(columns=['game_id', 'game_date', 'home_team', 'away_team'])
```

```python
nba_training_df = nba_training_df[nba_training_df['score_diff'] != 0]
nba_training_df['outcome'] = nba_training_df['score_diff'].apply(lambda x: 1 if x > 0 else 0)

y1 = nba_training_df['outcome']
y2 = nba_training_df['score_diff']
X1 = nba_training_df.drop(columns=['outcome', 'score_diff'])
X2 = nba_training_df.drop(columns=['outcome', 'score_diff'])

X1_train, X1_test, y1_train, y1_test = train_test_split(X1, y1, test_size=0.2, random_state=42)
X2_train, X2_test, y2_train, y2_test = train_test_split(X2, y2, test_size=0.2, random_state=42)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import cross_val_score, cross_validate, KFold
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score, classification_report, confusion_matrix
import numpy as np

# Initialize scaler for feature normalization
scaler = StandardScaler()

# Fit and transform the features
X1_scaled = scaler.fit_transform(X1)

print(f"Dataset shape: {X1_scaled.shape}")
print(f"Number of features: {X1_scaled.shape[1]}")
print(f"Number of samples: {X1_scaled.shape[0]}")
print(f"\nTarget distribution:")
print(y1.value_counts())
print(f"Positive class rate: {y1.mean():.4f}")
```

```python
# Initialize Logistic Regression model
log_reg = LogisticRegression(
    max_iter=1000,
    random_state=42,
    solver='lbfgs',
    class_weight='balanced'  # Handle any class imbalance
)

# Set up K-Fold cross-validation (k=5)
k_folds = 5
kf = KFold(n_splits=k_folds, shuffle=True, random_state=42)

# Define scoring metrics
scoring = {
    'accuracy': 'accuracy',
    'precision': 'precision',
    'recall': 'recall',
    'f1': 'f1',
    'roc_auc': 'roc_auc'
}

print(f"Performing {k_folds}-Fold Cross-Validation...")
print(f"{'='*60}\n")

# Perform cross-validation
cv_results = cross_validate(
    log_reg, 
    X1_scaled, 
    y1, 
    cv=kf, 
    scoring=scoring,
    return_train_score=True,
    n_jobs=-1  # Use all available cores
)

# Display results
print("Cross-Validation Results:")
print(f"{'='*60}")
print(f"\nTest Set Performance (mean ± std across {k_folds} folds):")
print(f"  Accuracy:  {cv_results['test_accuracy'].mean():.4f} ± {cv_results['test_accuracy'].std():.4f}")
print(f"  Precision: {cv_results['test_precision'].mean():.4f} ± {cv_results['test_precision'].std():.4f}")
print(f"  Recall:    {cv_results['test_recall'].mean():.4f} ± {cv_results['test_recall'].std():.4f}")
print(f"  F1 Score:  {cv_results['test_f1'].mean():.4f} ± {cv_results['test_f1'].std():.4f}")
print(f"  ROC AUC:   {cv_results['test_roc_auc'].mean():.4f} ± {cv_results['test_roc_auc'].std():.4f}")

print(f"\nTraining Set Performance (mean ± std across {k_folds} folds):")
print(f"  Accuracy:  {cv_results['train_accuracy'].mean():.4f} ± {cv_results['train_accuracy'].std():.4f}")
print(f"  Precision: {cv_results['train_precision'].mean():.4f} ± {cv_results['train_precision'].std():.4f}")
print(f"  Recall:    {cv_results['train_recall'].mean():.4f} ± {cv_results['train_recall'].std():.4f}")
print(f"  F1 Score:  {cv_results['train_f1'].mean():.4f} ± {cv_results['train_f1'].std():.4f}")
print(f"  ROC AUC:   {cv_results['train_roc_auc'].mean():.4f} ± {cv_results['train_roc_auc'].std():.4f}")

# Check for overfitting
print(f"\nOverfitting Check:")
train_test_gap = cv_results['train_accuracy'].mean() - cv_results['test_accuracy'].mean()
print(f"  Train-Test Accuracy Gap: {train_test_gap:.4f}")
if train_test_gap > 0.05:
    print("  ⚠️  Warning: Possible overfitting detected")
else:
    print("  ✓ Model generalizes well")

print(f"\n{'='*60}")
```

```python
# Train final model on all data and examine feature importance
print("Training final model on full dataset...")
log_reg_final = LogisticRegression(
    max_iter=1000,
    random_state=42,
    solver='lbfgs',
    class_weight='balanced'
)

log_reg_final.fit(X1_scaled, y1)

# Get feature importance (coefficients)
feature_importance = pd.DataFrame({
    'feature': X1.columns,
    'coefficient': log_reg_final.coef_[0],
    'abs_coefficient': np.abs(log_reg_final.coef_[0])
})

# Sort by absolute value
feature_importance = feature_importance.sort_values('abs_coefficient', ascending=False)

print(f"\nTop 20 Most Important Features:")
print(f"{'='*60}")
print(feature_importance.head(20).to_string(index=False))

print(f"\n\nModel Intercept: {log_reg_final.intercept_[0]:.4f}")
print(f"\n✓ Model training complete!")
```

```python
# Detailed fold-by-fold analysis
print("Fold-by-Fold Performance:")
print(f"{'='*60}\n")

for fold_idx in range(k_folds):
    print(f"Fold {fold_idx + 1}:")
    print(f"  Test Accuracy:  {cv_results['test_accuracy'][fold_idx]:.4f}")
    print(f"  Test Precision: {cv_results['test_precision'][fold_idx]:.4f}")
    print(f"  Test Recall:    {cv_results['test_recall'][fold_idx]:.4f}")
    print(f"  Test F1:        {cv_results['test_f1'][fold_idx]:.4f}")
    print(f"  Test ROC AUC:   {cv_results['test_roc_auc'][fold_idx]:.4f}")
    print()

# Create a summary DataFrame
fold_summary = pd.DataFrame({
    'Fold': range(1, k_folds + 1),
    'Accuracy': cv_results['test_accuracy'],
    'Precision': cv_results['test_precision'],
    'Recall': cv_results['test_recall'],
    'F1': cv_results['test_f1'],
    'ROC_AUC': cv_results['test_roc_auc']
})

print("\nSummary Table:")
print(fold_summary.to_string(index=False))
```

```python
# ============================================================
# Additional Models: Decision Tree, XGBoost, and Basic Neural Net
# ============================================================

from sklearn.tree import DecisionTreeClassifier
from sklearn.neural_network import MLPClassifier
from sklearn.pipeline import Pipeline
from sklearn.metrics import make_scorer

# (Optional) XGBoost
# If you don't have xgboost installed, uncomment the next line:
# !pip -q install xgboost
try:
    from xgboost import XGBClassifier
    _HAS_XGBOOST = True
except Exception as e:
    print("⚠️  xgboost not available in this environment. Install with: pip install xgboost")
    _HAS_XGBOOST = False

# We'll compare using the same CV splitter and scoring dict you already set up:
print(f"Using {k_folds}-fold CV with scoring: {list(scoring.keys())}")
```

```python
# Build model zoo
pos_rate = float(y1.mean())
neg_rate = 1.0 - pos_rate
# XGBoost's typical imbalance parameter (approx)
scale_pos_weight = (neg_rate / pos_rate) if pos_rate > 0 else 1.0

models = {}

# 1) Decision Tree (simple, interpretable baseline)
models["DecisionTree"] = DecisionTreeClassifier(
    random_state=42,
    class_weight="balanced",
    max_depth=None  # you can tune this later
)

# 2) XGBoost (strong baseline for tabular data)
if _HAS_XGBOOST:
    models["XGBoost"] = XGBClassifier(
        n_estimators=600,
        learning_rate=0.05,
        max_depth=4,
        subsample=0.9,
        colsample_bytree=0.9,
        reg_lambda=1.0,
        random_state=42,
        n_jobs=-1,
        eval_metric="logloss",
        scale_pos_weight=scale_pos_weight
    )

# 3) Basic Neural Net (MLP) — use a pipeline with scaling
models["MLP"] = Pipeline([
    ("scaler", StandardScaler()),
    ("mlp", MLPClassifier(
        hidden_layer_sizes=(128, 64),
        activation="relu",
        solver="adam",
        alpha=1e-4,
        batch_size=256,
        learning_rate_init=1e-3,
        max_iter=200,
        early_stopping=True,
        n_iter_no_change=10,
        random_state=42
    ))
])

# Evaluate each model with cross_validate
results = []
for name, model in models.items():
    print(f"\n{name}")
    print("-" * len(name))
    # NOTE:
    # - For Logistic Regression earlier you used X1_scaled.
    # - For tree/XGBoost, scaling isn't required, but it's fine to use raw X1.
    # - For MLP we embedded scaling into the pipeline, so use raw X1.
    cv = cross_validate(
        model,
        X1,      # raw features
        y1,
        cv=kf,
        scoring=scoring,
        return_train_score=True,
        n_jobs=-1
    )
    results.append({
        "model": name,
        "test_accuracy_mean": cv["test_accuracy"].mean(),
        "test_accuracy_std":  cv["test_accuracy"].std(),
        "test_precision_mean": cv["test_precision"].mean(),
        "test_precision_std":  cv["test_precision"].std(),
        "test_recall_mean": cv["test_recall"].mean(),
        "test_recall_std":  cv["test_recall"].std(),
        "test_f1_mean": cv["test_f1"].mean(),
        "test_f1_std":  cv["test_f1"].std(),
        "test_roc_auc_mean": cv["test_roc_auc"].mean(),
        "test_roc_auc_std":  cv["test_roc_auc"].std(),
        "train_accuracy_mean": cv["train_accuracy"].mean(),
        "train_accuracy_std":  cv["train_accuracy"].std(),
        "train_roc_auc_mean": cv["train_roc_auc"].mean(),
        "train_roc_auc_std":  cv["train_roc_auc"].std(),
        "train_test_acc_gap": cv["train_accuracy"].mean() - cv["test_accuracy"].mean(),
    })

results_df = pd.DataFrame(results)

# Rank by ROC AUC (usually good for probabilistic win/loss classification)
results_df = results_df.sort_values(by="test_roc_auc_mean", ascending=False)

# Pretty print
cols = [
    "model",
    "test_roc_auc_mean","test_roc_auc_std",
    "test_accuracy_mean","test_accuracy_std",
    "test_f1_mean","test_f1_std",
    "test_precision_mean","test_precision_std",
    "test_recall_mean","test_recall_std",
    "train_test_acc_gap"
]
display(results_df[cols].style.format({
    "test_roc_auc_mean":"{:.4f}","test_roc_auc_std":"{:.4f}",
    "test_accuracy_mean":"{:.4f}","test_accuracy_std":"{:.4f}",
    "test_f1_mean":"{:.4f}","test_f1_std":"{:.4f}",
    "test_precision_mean":"{:.4f}","test_precision_std":"{:.4f}",
    "test_recall_mean":"{:.4f}","test_recall_std":"{:.4f}",
    "train_test_acc_gap":"{:.4f}",
}))
```

```python
# Train the best-performing model on all data (by ROC AUC) and report a quick holdout metric too
best_name = results_df.iloc[0]["model"]
best_model = models[best_name]

print(f"Best model by CV ROC AUC: {best_name}")

# Holdout evaluation (same split you created earlier: X1_train/X1_test)
best_model.fit(X1_train, y1_train)

# Predict probabilities if available; else fallback
if hasattr(best_model, "predict_proba"):
    y_proba = best_model.predict_proba(X1_test)[:, 1]
    y_pred = (y_proba >= 0.5).astype(int)
else:
    y_pred = best_model.predict(X1_test)
    y_proba = None

from sklearn.metrics import accuracy_score, f1_score, roc_auc_score, precision_score, recall_score

print("\nHoldout Performance (20% split):")
print(f"  Accuracy : {accuracy_score(y1_test, y_pred):.4f}")
print(f"  Precision: {precision_score(y1_test, y_pred):.4f}")
print(f"  Recall   : {recall_score(y1_test, y_pred):.4f}")
print(f"  F1       : {f1_score(y1_test, y_pred):.4f}")
if y_proba is not None:
    print(f"  ROC AUC   : {roc_auc_score(y1_test, y_proba):.4f}")
else:
    print("  ROC AUC   : (no predict_proba available)")

# Feature importance for tree/XGBoost models (if supported)
if best_name in ["DecisionTree", "XGBoost"]:
    if best_name == "DecisionTree":
        importances = best_model.feature_importances_
    else:
        # XGBClassifier exposes feature_importances_
        importances = best_model.feature_importances_

    fi = pd.DataFrame({"feature": X1.columns, "importance": importances}).sort_values("importance", ascending=False)
    print("\nTop 25 Feature Importances:")
    display(fi.head(25))
```

```python
# XGBoost Hyperparameter Grid Search
print("="*60)
print("XGBoost Hyperparameter Grid Search")
print("="*60)

from sklearn.model_selection import GridSearchCV
import xgboost as xgb

# Define parameter grid for XGBoost
param_grid = {
    'n_estimators': [100, 200, 300],
    'max_depth': [3, 5, 7, 10],
    'learning_rate': [0.01, 0.05, 0.1, 0.2],
    'min_child_weight': [1, 3, 5],
    'subsample': [0.6, 0.8, 1.0],
    'colsample_bytree': [0.6, 0.8, 1.0],
    'gamma': [0, 0.1, 0.2]
}

print(f"\nParameter grid:")
for param, values in param_grid.items():
    print(f"  {param}: {values}")

print(f"\nTotal combinations: {np.prod([len(v) for v in param_grid.values()])}")

# Initialize XGBoost classifier
xgb_clf = xgb.XGBClassifier(
    objective='binary:logistic',
    eval_metric='logloss',
    random_state=42,
    n_jobs=-1
)

# Initialize GridSearchCV
grid_search = GridSearchCV(
    estimator=xgb_clf,
    param_grid=param_grid,
    scoring='roc_auc',
    cv=5,
    verbose=2,
    n_jobs=-1,
    return_train_score=True
)

print("\nStarting grid search (this may take a while)...")
print("Using 5-fold cross-validation with ROC AUC scoring\n")

# Fit grid search
grid_search.fit(X1_train, y1_train)

print("\n" + "="*60)
print("Grid Search Complete!")
print("="*60)

# Best parameters
print("\nBest Parameters:")
for param, value in grid_search.best_params_.items():
    print(f"  {param}: {value}")

print(f"\nBest CV ROC AUC Score: {grid_search.best_score_:.4f}")

# Test on holdout set with best model
best_xgb = grid_search.best_estimator_
y_pred_best = best_xgb.predict(X1_test)
y_proba_best = best_xgb.predict_proba(X1_test)[:, 1]

print("\nHoldout Performance (Best XGBoost):")
print(f"  Accuracy : {accuracy_score(y1_test, y_pred_best):.4f}")
print(f"  Precision: {precision_score(y1_test, y_pred_best):.4f}")
print(f"  Recall   : {recall_score(y1_test, y_pred_best):.4f}")
print(f"  F1       : {f1_score(y1_test, y_pred_best):.4f}")
print(f"  ROC AUC   : {roc_auc_score(y1_test, y_proba_best):.4f}")

# Feature importance
importances_best = best_xgb.feature_importances_
fi_best = pd.DataFrame({
    "feature": X1.columns, 
    "importance": importances_best
}).sort_values("importance", ascending=False)

print("\nTop 25 Feature Importances (Best Model):")
display(fi_best.head(25))

# Show top 10 parameter combinations
print("\nTop 10 Parameter Combinations by CV Score:")
results_df = pd.DataFrame(grid_search.cv_results_)
results_df = results_df.sort_values('mean_test_score', ascending=False)
display(results_df[['params', 'mean_test_score', 'std_test_score', 'mean_train_score']].head(10))
```

```python
# Save the best model to nba/models
import os
import pickle
import json
from datetime import datetime

# Create models directory if it doesn't exist
os.makedirs("models", exist_ok=True)

# Save the best model using pickle
model_path = "models/xgboost_best_model.pkl"
with open(model_path, 'wb') as f:
    pickle.dump(best_xgb, f)
print(f"\n✓ Best model saved to {model_path}")

# Save best parameters and performance metrics
model_info = {
    "model_type": "XGBoost",
    "best_params": grid_search.best_params_,
    "cv_roc_auc": float(grid_search.best_score_),
    "holdout_metrics": {
        "accuracy": float(accuracy_score(y1_test, y_pred_best)),
        "precision": float(precision_score(y1_test, y_pred_best)),
        "recall": float(recall_score(y1_test, y_pred_best)),
        "f1": float(f1_score(y1_test, y_pred_best)),
        "roc_auc": float(roc_auc_score(y1_test, y_proba_best))
    },
    "training_date": datetime.now().strftime("%Y-%m-%d %H:%M:%S"),
    "training_samples": len(X1_train),
    "test_samples": len(X1_test),
    "features": list(X1.columns)
}

info_path = "models/xgboost_best_model_info.json"
with open(info_path, 'w') as f:
    json.dump(model_info, f, indent=2)
print(f"✓ Model info saved to {info_path}")

# Also save feature importances
fi_path = "models/xgboost_feature_importances.csv"
fi_best.to_csv(fi_path, index=False)
print(f"✓ Feature importances saved to {fi_path}")

print(f"\nModel artifacts saved:")
print(f"  1. {model_path} - Trained model (pickle)")
print(f"  2. {info_path} - Model metadata (JSON)")
print(f"  3. {fi_path} - Feature importances (CSV)")
```
