```python
import pandas as pd
import os
from pathlib import Path

# Read games.csv to get all available game_ids
games_df = pd.read_csv('data/games.csv', dtype={"game_id": str})
games_ids = set(games_df['game_id'].astype(str))

# Read all schedule files and collect game_ids
schedules_dir = Path('data/schedules')
all_schedule_game_ids = set()
schedule_files = sorted(schedules_dir.glob('schedule_*.csv'))

for schedule_file in schedule_files:
    df = pd.read_csv(schedule_file, dtype={"GAME_ID": str}, usecols=["GAME_ID", "GAME_DATE", "MATCHUP"])
    season_game_ids = set(df['GAME_ID'].astype(str))
    all_schedule_game_ids.update(season_game_ids)
    # print(f"  {schedule_file.name}: {len(season_game_ids)} games")

print(f"\nTotal unique games in all schedules: {len(all_schedule_game_ids)}")

# Find missing game_ids
missing_from_games = all_schedule_game_ids - games_ids
print(f"\n{'✓ PASS' if len(missing_from_games) == 0 else '✗ FAIL'}: Games missing from games.csv: {len(missing_from_games)}")

if missing_from_games:
    # print("\nMissing game IDs (first 20):")
    # for gid in sorted(missing_from_games)[:20]:
    #     print(f"  - {gid}")
    # if len(missing_from_games) > 20:
    #     print(f"  ... and {len(missing_from_games) - 20} more")
    
    # Show which seasons have missing games
    print("\n" + "=" * 80)
    print("MISSING GAMES IN games.csv BREAKDOWN BY SCHEDULE:")
    print("=" * 80)
    print(f"{'Schedule':<30} {'Total Games':<15} {'Missing':<15} {'% Missing':<15}")
    print("-" * 80)
    
    total_games_sum = 0
    total_missing_sum = 0
    
    for schedule_file in schedule_files:
        df = pd.read_csv(schedule_file, dtype={"GAME_ID": str}, usecols=["GAME_ID"])
        season_game_ids = set(df['GAME_ID'].astype(str))
        season_missing = season_game_ids - games_ids
        total_games = len(season_game_ids)
        missing_count = len(season_missing)
        pct_missing = (missing_count / total_games * 100) if total_games > 0 else 0
        
        total_games_sum += total_games
        total_missing_sum += missing_count
        
        status = "✗" if missing_count > 0 else "✓"
        print(f"{status} {schedule_file.name:<28} {total_games:<15} {missing_count:<15} {pct_missing:>6.2f}%")
    
    print("-" * 80)
    total_pct = (total_missing_sum / total_games_sum * 100) if total_games_sum > 0 else 0
    print(f"  {'TOTAL':<28} {total_games_sum:<15} {total_missing_sum:<15} {total_pct:>6.2f}%")
    print("=" * 80)
else:
    print("  All schedule game IDs are present in games.csv!")

print("\n" + "=" * 80)
```

```python
import pandas as pd
import os
from pathlib import Path

# Check if all game_ids from schedules have corresponding files in players_live/
print("=" * 80)
print("CHECKING: Game IDs from schedules have files in players_live/")
print("=" * 80)

# Get all player live files
players_live_dir = Path('data/players_live')
player_files = list(players_live_dir.glob('*.csv'))

# Extract game_ids from filenames (format: GAMEID_TEAM1_TEAM2.csv)
players_live_game_ids = set()
for file in player_files:
    game_id = file.stem.split('_')[0]  # Get first part before underscore
    players_live_game_ids.add(game_id)

print(f"Total unique game IDs in players_live/: {len(players_live_game_ids)}")
print(f"Total player files: {len(player_files)}")

# Read all schedule files and collect game_ids
schedules_dir = Path('data/schedules')
all_schedule_game_ids = set()
schedule_files = sorted(schedules_dir.glob('schedule_*.csv'))

for schedule_file in schedule_files:
    df = pd.read_csv(schedule_file, dtype={"GAME_ID": str}, usecols=["GAME_ID", "GAME_DATE", "MATCHUP"])
    season_game_ids = set(df['GAME_ID'].astype(str))
    all_schedule_game_ids.update(season_game_ids)
    print(f"  {schedule_file.name}: {len(season_game_ids)} games")

print(f"\nTotal unique games in all schedules: {len(all_schedule_game_ids)}")

# Find missing player files
missing_from_players = all_schedule_game_ids - players_live_game_ids
print(f"\n{'✓ PASS' if len(missing_from_players) == 0 else '✗ FAIL'}: Games missing from players_live/: {len(missing_from_players)}")

if missing_from_players:
    print("\nMissing game IDs (first 20):")
    for gid in sorted(missing_from_players)[:20]:
        print(f"  - {gid}")
    if len(missing_from_players) > 20:
        print(f"  ... and {len(missing_from_players) - 20} more")
    
    # Show which seasons have missing player files
    print("\n" + "=" * 80)
    print("MISSING PLAYER FILES BREAKDOWN BY SCHEDULE:")
    print("=" * 80)
    print(f"{'Schedule':<30} {'Total Games':<15} {'Missing':<15} {'% Missing':<15}")
    print("-" * 80)
    
    total_games_sum = 0
    total_missing_sum = 0
    
    for schedule_file in schedule_files:
        df = pd.read_csv(schedule_file, dtype={"GAME_ID": str}, usecols=["GAME_ID"])
        season_game_ids = set(df['GAME_ID'].astype(str))
        season_missing = season_game_ids - players_live_game_ids
        total_games = len(season_game_ids)
        missing_count = len(season_missing)
        pct_missing = (missing_count / total_games * 100) if total_games > 0 else 0
        
        total_games_sum += total_games
        total_missing_sum += missing_count
        
        status = "✗" if missing_count > 0 else "✓"
        print(f"{status} {schedule_file.name:<28} {total_games:<15} {missing_count:<15} {pct_missing:>6.2f}%")
    
    print("-" * 80)
    total_pct = (total_missing_sum / total_games_sum * 100) if total_games_sum > 0 else 0
    print(f"  {'TOTAL':<28} {total_games_sum:<15} {total_missing_sum:<15} {total_pct:>6.2f}%")
    print("=" * 80)
else:
    print("  All schedule game IDs have corresponding player files!")

# Extra: Check for player files not in schedules (orphaned files)
extra_player_files = players_live_game_ids - all_schedule_game_ids
if extra_player_files:
    print(f"\nNote: Found {len(extra_player_files)} game IDs in players_live/ that are NOT in schedules")
    print("(These might be from the 2000-01 season which has no schedule data, or playoff/special games)")

print("\n" + "=" * 80)
```
